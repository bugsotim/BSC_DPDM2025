```python
import os
import glob
```

```python
# 2. กำหนด Path หลักที่เก็บโฟลเดอร์รายเดือน
base_path = '/content/drive/MyDrive/BSC DPDM midterm 2025/Input - (Operation Daily rain from API_ 2025)'

# 3. สร้างฟังก์ชันสำหรับรวมไฟล์รายวันในแต่ละโฟลเดอร์
def merge_daily_files(month_folder):
    full_path = os.path.join(base_path, month_folder)
    # ดึงไฟล์ .csv ทั้งหมดและเรียงลำดับวันที่ 01-31
    all_files = sorted(glob.glob(os.path.join(full_path, "*.csv")))

    if not all_files:
        print(f"⚠️ ไม่พบไฟล์ในโฟลเดอร์: {month_folder}")
        return pd.DataFrame()

    print(f"กำลังรวมไฟล์เดือน {month_folder}: พบทั้งหมด {len(all_files)} ไฟล์")

    # อ่านไฟล์รายวันทั้งหมดมาเก็บใน List
    temp_list = [pd.read_csv(f) for f in all_files]

    # รวม (Concatenate) เป็นตารางเดียว
    return pd.concat(temp_list, ignore_index=True)

# 4. ดำเนินการสร้างตัวแปร df_07 ถึง df_12
df_07 = merge_daily_files('2025-07')
df_08 = merge_daily_files('2025-08')
df_09 = merge_daily_files('2025-09')
df_10 = merge_daily_files('2025-10')
df_11 = merge_daily_files('2025-11')
df_12 = merge_daily_files('2025-12')

# 5. แสดงโครงสร้างข้อมูลและอธิบายความหมาย (ตัวอย่างจาก df_07)
print("\n" + "="*30)
print("สรุปโครงสร้างข้อมูลรายเดือน (ตัวอย่างเดือน 7)")
print("="*30)
df_07.info()
```

### ข้อ 2 กรองข้อมูลให้เหลือเฉพาะสถานีใน Stations in the model_Hll.csv แสดงจำนวนสถานีก่อน/หลังกรอง

```python
# Prompt : กรองข้อมูลให้เหลือเฉพาะสถานีใน Stations in the model_Hll.csv แสดงจำนวนสถานีก่อน/หลังกรอง
# 1. โหลดไฟล์ Stations in the model_Hll.csv
stations_model_path = '/content/drive/MyDrive/BSC DPDM midterm 2025/Input - (Operation Daily rain from API_ 2025)/Stations in the model_HII.csv'
stations_in_model_df = pd.read_csv(stations_model_path)

# 2. ดึงรายการรหัสสถานีที่ต้องการกรอง
# สังเกตจากไฟล์ output และ hii_daily_rain คอลัมน์รหัสสถานีมักจะชื่อ 'Node' หรือ 'tele_station_oldcode'
# ตรวจสอบ Stations in the model_Hll.csv เพื่อดูชื่อคอลัมน์รหัสสถานีที่ถูกต้อง
# สมมติว่าคอลัมน์ชื่อ 'Node' หรือ 'station_code' หรือ 'tele_station_oldcode' ใน Stations in the model_Hll.csv
# ลองใช้ 'station_code' ก่อน หากผิดพลาดจะต้องปรับแก้ตามโครงสร้างจริงของ Stations in the model_Hll.csv

# สำหรับจุดประสงค์นี้ เราจะใช้ 'tele_station_oldcode' จากตัวอย่าง df_07.head() และ df_12.head()
# และสันนิษฐานว่า Stations in the model_Hll.csv มีคอลัมน์ 'Node' หรือ 'station_code' ที่ตรงกับ 'tele_station_oldcode'

# ดึงคอลัมน์ 'tele_station_oldcode' จาก dataframe ที่โหลดมา
# หากชื่อคอลัมน์ใน Stations in the model_Hll.csv ไม่ใช่ 'tele_station_oldcode' ต้องปรับแก้ตรงนี้
# ตัวอย่างเช่น ถ้าเป็น 'Node' ก็ใช้ stations_to_filter = stations_in_model_df['Node'].unique()
# หรือถ้าเป็น 'station_code' ก็ใช้ stations_to_filter = stations_in_model_df['station_code'].unique()

# ตรวจสอบชื่อคอลัมน์ใน Stations in the model_Hll.csv ก่อนใช้งานจริง
# print(stations_in_model_df.head())

# สำหรับตอนนี้สมมติว่าคอลัมน์ที่มี station code คือ 'tele_station_oldcode'
# แต่จากที่เห็นในไฟล์ Stations in the model_Hll.csv น่าจะเป็นคอลัมน์ที่ชื่อว่า 'Node'
stations_to_filter = stations_in_model_df['Node'].unique()

print(f"จำนวนสถานีทั้งหมดในไฟล์ model: {len(stations_to_filter)} สถานี")

# 3. เตรียมรายการ DataFrame รายเดือน
monthly_dfs = {
    'df_07': df_07,
    'df_08': df_08,
    'df_09': df_09,
    'df_10': df_10,
    'df_11': df_11,
    'df_12': df_12
}

filtered_monthly_dfs = {}

print("\n=== ผลการกรองข้อมูลรายเดือน ===")
for month_name, df in monthly_dfs.items():
    if df.empty:
        print(f"--- {month_name}: DataFrame ว่างเปล่า ไม่สามารถกรองได้ ---")
        continue

    # นับจำนวนสถานีก่อนกรอง
    stations_before_filter = df['tele_station_oldcode'].nunique()

    # กรองข้อมูล
    df_filtered = df[df['tele_station_oldcode'].isin(stations_to_filter)].copy()
    filtered_monthly_dfs[month_name] = df_filtered

    # นับจำนวนสถานีหลังกรอง
    stations_after_filter = df_filtered['tele_station_oldcode'].nunique()

    print(f"--- {month_name} ---")
    print(f"จำนวนสถานีก่อนกรอง: {stations_before_filter} สถานี")
    print(f"จำนวนสถานีหลังกรอง: {stations_after_filter} สถานี")
    print(f"จำนวนแถวหลังกรอง: {len(df_filtered)} แถว")

    # แสดงตัวอย่างข้อมูลที่ถูกกรอง (ไม่กี่แถวแรก)
    # if not df_filtered.empty:
    #     display(df_filtered.head())
    # else:
    #     print("    (ไม่มีข้อมูลสถานีที่ตรงตามเงื่อนไขในเดือนนี้)")
```

### ข้อ 3 ระบุข้อมูลสูญหาย ( recordที่หายไป ) ของแต่ละสถานี แสดงจำนวนและสัดส่วน

```python
# Prompt : ระบุข้อมูลสูญหาย ( recordที่หายไป ) ของแต่ละสถานี แสดงจำนวนและสัดส่วนโดยข้อมุลที่สูญหายในที่นี้หมายถึง record ที่ไม่มีไฟล์เลยไม่ใช่ค่า Null หรือ NaN) ตามขั้นตอนของ Data Processing
# 1. สร้างช่วงเวลาที่ควรจะมีจริง (Expected Timeline)
# เราจะสร้างช่วงตั้งแต่วันที่ 1 ก.ค. ถึง 31 ธ.ค. 2025
expected_dates = pd.date_range(start='2025-07-01', end='2025-12-31', freq='D')
total_expected_days = len(expected_dates)

missing_summary = []

print(f"--- เริ่มการวิเคราะห์ข้อมูลสูญหาย (ช่วงวันที่ตรวจสอบ: 2025-07-01 ถึง 2025-12-31) ---")
print(f"จำนวนวันที่ควรจะมีทั้งหมด: {total_expected_days} วัน\n")

# 2. รวม DataFrame ทุกเดือนที่กรองแล้วเข้าด้วยกันเป็นก้อนเดียวเพื่อให้ตรวจสอบง่าย
df_all_half_year = pd.concat(filtered_monthly_dfs.values(), ignore_index=True)
df_all_half_year['measure_datetime'] = pd.to_datetime(df_all_half_year['measure_datetime'])

# 3. วนลูปตรวจสอบทีละสถานี (ตามรายชื่อใน Model)
for node in stations_to_filter:
    # ดึงข้อมูลจริงของสถานีนี้
    station_actual_data = df_all_half_year[df_all_half_year['tele_station_oldcode'] == node]

    # นับจำนวนวันที่ปรากฏจริง
    actual_days_count = station_actual_data['measure_datetime'].nunique()

    # คำนวณจำนวน Record ที่หายไป
    missing_records = total_expected_days - actual_days_count

    # คำนวณสัดส่วน (Percentage of Missingness)
    missing_ratio = (missing_records / total_expected_days) * 100

    # บันทึกผลลัพธ์
    missing_summary.append({
        'Station_Node': node,
        'Expected_Records': total_expected_days,
        'Actual_Records': actual_days_count,
        'Missing_Records': missing_records,
        'Missing_Ratio (%)': round(missing_ratio, 2)
    })

# 4. แปลงเป็น DataFrame เพื่อสรุปผลตามหลัก Data Mining
df_missing_report = pd.DataFrame(missing_summary)

# 5. แสดงผลลัพธ์เฉพาะสถานีที่มีปัญหา (Missing > 0) เรียงจากมากไปน้อย
df_problem_stations = df_missing_report[df_missing_report['Missing_Records'] > 0].sort_values(by='Missing_Records', ascending=False)

print("=== สรุปผลสถานีที่มีข้อมูลสูญหาย ===")
if not df_problem_stations.empty:
    display(df_problem_stations)
else:
    print("ไม่พบข้อมูลสูญหายในทุกสถานี (ข้อมูลครบ 100%)")
```

### ข้อ 4 คัดสถานีที่มีข้อมูลสูญหาย >= 20% ออก แสดงรายชื่อสถานีที่ถูกคัดออกและจำนวนที่เหลือ

```python
# Prompt : ช่วยเขียนโค้ดที่ คัดสถานีที่มีข้อมูลสูญหาย >= 20% ออก แสดงรายชื่อสถานีที่ถูกคัดออกและจำนวนที่เหลือ
# 1. กำหนดเกณฑ์ในการคัดออก (Threshold)
threshold_percent = 20.0

# 2. แยกสถานีที่ผ่านเกณฑ์ และสถานีที่ถูกคัดออก
# อ้างอิงจาก df_missing_report ที่เราสร้างไว้ในขั้นตอนก่อนหน้า
dropped_stations_df = df_missing_report[df_missing_report['Missing_Ratio (%)'] >= threshold_percent]
kept_stations_df = df_missing_report[df_missing_report['Missing_Ratio (%)'] < threshold_percent]

# 3. ดึงรายชื่อ Node ของสถานีที่ได้ไปต่อ
valid_station_nodes = kept_stations_df['Station_Node'].tolist()

# 4. แสดงผลการคัดกรอง
print(f"=== รายงานการ Data Cleaning (Threshold: Missing < {threshold_percent}%) ===")
print(f"1. จำนวนสถานีเริ่มต้น: {len(df_missing_report)} สถานี")
print(f"2. จำนวนสถานีที่ถูกคัดออก: {len(dropped_stations_df)} สถานี")
print(f"3. จำนวนสถานีที่คงเหลือ: {len(kept_stations_df)} สถานี")

if not dropped_stations_df.empty:
    print("\n--- รายชื่อสถานีที่ถูกคัดออก (Missing >= 20%) ---")
    display(dropped_stations_df[['Station_Node', 'Missing_Records', 'Missing_Ratio (%)']].sort_values(by='Missing_Ratio (%)', ascending=False))
else:
    print("\n--- ไม่มีสถานีใดถูกคัดออก (ทุกสถานีผ่านเกณฑ์) ---")

# 5. กรองข้อมูลใน DataFrame หลักให้เหลือเฉพาะสถานีที่ผ่านเกณฑ์
# (ตัวอย่างการกรองข้อมูลรวมครึ่งปีหลัง)
df_final_clean = df_all_half_year[df_all_half_year['tele_station_oldcode'].isin(valid_station_nodes)]

print("\n" + "="*50)
print(f"จำนวน Record รวมใน Database หลัง Cleaning: {len(df_final_clean)} แถว")
```

### ข้อ 5  เติมข้อมูลที่หายไปด้วยค่า median ตาม Day of Year จาก historical แยกตามสถานี ตรวจสอบว่าไม่มี record หายอีก

```python
# Prompt : ข้อ 5 เติมข้อมูลที่หายไปด้วยค่า median ตาม Day of Year จาก historical แยกตามสถานี ตรวจสอบว่าไม่มี record หายอีก
import numpy as np

# 1. เตรียม Master Calendar สำหรับทุกสถานีที่ผ่านเกณฑ์ (เพื่อให้แน่ใจว่า Record จะไม่หายอีก)
valid_nodes = kept_stations_df['Station_Node'].unique()
expected_dates = pd.date_range(start='2025-07-01', end='2025-12-31', freq='D')

# สร้าง DataFrame แม่แบบที่มีทุกสถานี x ทุกวัน
master_grid = pd.MultiIndex.from_product([valid_nodes, expected_dates], names=['tele_station_oldcode', 'measure_datetime']).to_frame(index=False)
master_grid['measure_datetime'] = pd.to_datetime(master_grid['measure_datetime'])

# 2. นำข้อมูลที่สะอาดแล้ว (df_final_clean) มา Merge เข้ากับ Master Grid
# วันที่ไม่มีข้อมูลจะกลายเป็น NaN โดยอัตโนมัติเพื่อให้เราเติมค่า
df_imputed = pd.merge(master_grid, df_final_clean, on=['tele_station_oldcode', 'measure_datetime'], how='left')

# 3. สร้างคอลัมน์ 'day_of_year' เพื่อใช้หาค่า Median ตามฤดูกาล
df_imputed['day_of_year'] = df_imputed['measure_datetime'].dt.dayofyear

# 4. คำนวณค่า Median แยกตามสถานี และตามวันในรอบปี (Historical Median Imputation)
# เราจะใช้ค่าที่มีอยู่มาหาค่ากลางของแต่ละสถานีในแต่ละวัน
station_doy_median = df_imputed.groupby(['tele_station_oldcode', 'day_of_year'])['rainfall_value'].transform('median')

# 5. เติมค่า Median ลงในช่องว่าง (เฉพาะ Record ที่เป็น NaN)
df_imputed['rainfall_value'] = df_imputed['rainfall_value'].fillna(station_doy_median)

# 6. ตรวจสอบครั้งสุดท้าย: หากยังมี NaN (กรณีวันที่นั้นไม่มีข้อมูลเลยในทุกสถานี) ให้เติมด้วย 0 หรือ Median รวม
df_imputed['rainfall_value'] = df_imputed['rainfall_value'].fillna(0)

# 7. ตรวจสอบความสมบูรณ์ (Verification)
print("=== ตรวจสอบความสมบูรณ์หลังการ Imputation ===")
total_records = len(df_imputed)
missing_after = df_imputed['rainfall_value'].isnull().sum()
expected_total = len(valid_nodes) * len(expected_dates)

print(f"1. จำนวน Record ที่ควรจะมี (Stations x Days): {expected_total}")
print(f"2. จำนวน Record ที่มีจริงหลัง Imputation: {total_records}")
print(f"3. จำนวนค่าที่เป็น Null/NaN: {missing_after}")

if total_records == expected_total and missing_after == 0:
    print("\n✅ สำเร็จ: ไม่มี Record หายและไม่มีค่าว่างในฐานข้อมูลแล้ว!")

# แสดงตัวอย่างข้อมูลที่เติมแล้ว
display(df_imputed.head())
```

### ข้อ 6 อภิปราย : เหตุใดใช้ median แทน Mean

การเลือกใช้ค่ามัธยฐาน (Median) แทนค่าเฉลี่ย (Mean) ในกระบวนการเตรียมข้อมูลปริมาณน้ำฝนตามหลัก Data Preprocessing นั้น มีเหตุผลสำคัญอยู่ที่ลักษณะทางสถิติของข้อมูลธรรมชาติที่มีความแปรปรวนสูง โดยเฉพาะปริมาณน้ำฝนที่มีการกระจายตัวแบบเบ้ขวา (Skewed Distribution) ซึ่งหมายความว่าข้อมูลส่วนใหญ่จะมีค่าเป็นศูนย์หรือน้อยมาก แต่จะมีบางวันที่เกิดฝนตกหนักผิดปกติจนกลายเป็นค่าผิดปกติ (Outliers) หากเราเลือกใช้ค่าเฉลี่ยในการเติมข้อมูล ข้อมูลที่สูงผิดปกติเพียงไม่กี่วันจะดึงค่าเฉลี่ยให้สูงขึ้นเกินความเป็นจริง ส่งผลให้ค่าที่เติมลงไปในวันที่ข้อมูลสูญหายนั้นมีค่า "เปียก" หรือสูงกว่าสภาพอากาศที่ควรจะเป็นในเชิงสถิติ

ในทางกลับกัน ค่ามัธยฐานมีความทนทานต่อค่าผิดปกติ (Robustness) มากกว่า เพราะเป็นการพิจารณาตำแหน่งกึ่งกลางของชุดข้อมูล ทำให้ค่าที่ได้สะท้อนถึงเหตุการณ์ที่เกิดขึ้นบ่อยที่สุดในประวัติศาสตร์ของวันนั้นๆ ได้อย่างสมเหตุสมผล การใช้มัธยฐานตามวันในรอบปี (Day of Year) จึงช่วยรักษาลักษณะทางกายภาพของข้อมูลให้สอดคล้องกับฤดูกาลจริง ทำให้ข้อมูลที่ได้รับการเติมเต็มมีความน่าเชื่อถือและไม่เกิดความลำเอียง (Bias) เมื่อนำไปใช้ในขั้นตอนการทำเหมืองข้อมูลหรือสร้างแบบจำลองพยากรณ์ในลำดับถัดไป

### ข้อ 7 คำนวณผลรวมปริมาณน้ำฝนรายเดือน จัดรุปแบบให้ตรงกับ Output_monthly_rainfall.csv

```python
import pandas as pd

# 1. จัดกลุ่มข้อมูลและหาผลรวมรายเดือนจากข้อมูลที่เติมค่าเรียบร้อยแล้ว (df_imputed)
monthly_summary = df_imputed.groupby(['tele_station_oldcode', 'month'])['rainfall_value'].sum().reset_index()

# 2. ทำการ Pivot เพื่อเปลี่ยนจากแนวตั้งเป็นแนวนอน (Wide Format)
output_monthly = monthly_summary.pivot(index='tele_station_oldcode', columns='month', values='rainfall_value')

# 3. ตั้งชื่อคอลัมน์ใหม่ให้ตรงตามรูปแบบ Month_XX
output_monthly.columns = [f'Month_{int(col):02d}' for col in output_monthly.columns]

# 4. เพิ่มคอลัมน์ผลรวมน้ำฝนสะสมตลอด 6 เดือน
output_monthly['Total_Rainfall'] = output_monthly.sum(axis=1)

# 5. รีเซ็ต Index และเปลี่ยนชื่อคอลัมน์รหัสสถานีให้เป็น 'Node'
output_monthly = output_monthly.reset_index().rename(columns={'tele_station_oldcode': 'Node'})

# 6. สร้างไฟล์ CSV ลงใน Local Storage ของเครื่องเซิร์ฟเวอร์ (ไม่ต้องระบุ Path Drive)
file_name = 'Output_monthly_rainfall.csv'
output_monthly.to_csv(file_name, index=False)

# 7. แสดงตัวอย่างตารางผลลัพธ์
print(f"--- สร้างไฟล์ {file_name} สำเร็จ ---")
display(output_monthly.head())
```
